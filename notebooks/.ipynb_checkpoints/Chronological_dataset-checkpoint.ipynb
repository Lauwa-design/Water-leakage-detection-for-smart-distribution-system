```python
import pandas as pd
import json
from pathlib import Path

base_path = Path("scenarios_2000")   # folder with all 2000 scenarios
all_data = []

for scenario in base_path.iterdir():

    pressures = pd.read_csv(scenario / "pressures.csv")
    flows = pd.read_csv(scenario / "flows.csv")
    demands = pd.read_csv(scenario / "demands.csv")
    labels = pd.read_csv(scenario / "labels.csv")

    # Merge datasets on time
    df = pressures.merge(flows, on="time")
    df = df.merge(demands, on="time")
    df = df.merge(labels, on="time")

    # Add scenario ID
    df["scenario_id"] = scenario.name

    # Load leak metadata
    with open(scenario / "leak_type.json") as f:
        leak_type = json.load(f)["type"]

    df["leak_type"] = leak_type

    all_data.append(df)

# Combine all scenarios
dataset = pd.concat(all_data)

# Sort chronologically
dataset = dataset.sort_values(["scenario_id", "time"])

dataset.to_csv("chronological_dataset.csv", index=False)
```
